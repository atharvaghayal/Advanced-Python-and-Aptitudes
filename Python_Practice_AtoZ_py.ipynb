## **Control statements** in Python are used to manage the flow of execution in a program. They help in decision-making, looping, and altering the normal sequence of execution.

**Conditional Statements** (if, elif, else) – These are used to execute a block of code based on specific conditions.

**Looping Statements** (for, while) – These allow repeated execution of a block of code until a condition is met.

**Jump Statements** (break, continue, pass) – break exits a loop, continue skips the current iteration, and pass acts as a placeholder.

These statements help in writing efficient and structured programs.

```python
# Control Statements :

# Conditional Statements :

age = int(input("Enter your age: "))

if age < 18:
    print("You are a minor.")
elif 18 <= age < 60:
    print("You are an adult.")
else:
    print("You are a senior citizen.")

# Looping Statement(using for loop) :

for i in range(1, 6):  # Loops from 1 to 5
    print(f"Number: {i}")

#Looping Statement(using while loop) :

count = 1
while count <= 5:
    print(f"Count: {count}")
    count += 1

# Jump Statements (break,pass,continue) :

for num in range(1, 6):
    if num == 3:
        continue  # Skips 3
    elif num == 5:
        break  # Stops loop at 5
    else:
        pass  # Placeholder (does nothing)
    print(num)

# All (Control statement,Looping Statement, Jump Statement Combined Example) :

for i in range(1, 11):
    if i % 2 == 0:
        continue  # Skips even numbers
    elif i == 9:
        break  # Stops loop when `i` is 9
    else:
        print(f"Odd Number: {i}")  # Prints only odd numbers except 9
```

```python
# Fully Combined Example :

import random

# Generate a random number between 1 and 20
secret_number = random.randint(1, 20)
attempts = 0

print("Welcome to the Number Guessing Game! (Guess between 1-20)")

# Using a for loop to allow a maximum of 5 attempts
for attempt in range(1, 6):
    guess = int(input(f"Attempt {attempt}: Enter your guess: "))
    attempts += 1

    if guess < secret_number:
        print("Too low! Try again.")
        continue  # Skips the remaining part of the loop and starts next iteration
    elif guess > secret_number:
        print("Too high! Try again.")
        pass  # Does nothing, just a placeholder
    else:
        print(f"🎉 Congratulations! You guessed it in {attempts} attempts.")
        break  # Exits loop if guessed correctly

# If the user couldn't guess in 5 attempts, allow them an extra chance using `while` loop
else:
    print("\nOops! You've used all 5 attempts.")

    # Asking for an extra attempt
    while True:
        choice = input("Do you want one last chance? (yes/no): ").lower()

        if choice == "yes":
            extra_guess = int(input("Enter your final guess: "))

            if extra_guess == secret_number:
                print(f"🎉 Great! You guessed it right in the extra attempt!")
            else:
                print(f"❌ Sorry! The correct number was {secret_number}.")
            break  # Exits while loop
        elif choice == "no":
            print(f"The correct number was {secret_number}. Better luck next time!")
            break
        else:
            print("Invalid input, please type 'yes' or 'no'.")
```

# Star and Number Patterns in Python :

Pattern printing is one of the fundamental concepts in Python, often used to develop logic-building skills for beginners. It involves using nested loops (one loop controlling rows and another controlling columns) to display various patterns such as stars (*), numbers, alphabets, or special characters. These patterns help in understanding loops, conditionals, and iterations, which are essential for programming logic.

Star and number patterns are widely used in competitive programming, logic development, and technical interviews to test problem-solving ability. They also find applications in graphics programming, ASCII art, and mathematical representations in Python. Below are Python programs for all major types of star patterns and number patterns.

```python
# Python Programs for Star Patterns...

# 1. for Simple Right-Angled Triangle Star Pattern..
n = 5
for i in range(1, n + 1):
    print("* " * i)

# 2.Inverted Right-Angled Triangle Star Pattern.
n = 5
for i in range(n, 0, -1):
    print("* " * i)

# 3. Pyramid Star Pattern.
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)

# 4. Inverted Pyramid Star Pattern.
n = 5
for i in range(n, 0, -1):
    print(" " * (n - i) + "* " * i)

# 5. Diamond Star Pattern
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " * i)

# 6. Hollow Diamond Star Pattern.
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "* " if i == 1 else " " * (n - i) + "* " + "  " * (i - 2) + "* ")
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " if i == 1 else " " * (n - i) + "* " + "  " * (i - 2) + "* ")

# 7. Hollow Square Star Pattern.
n = 5
for i in range(n):
    if i == 0 or i == n - 1:
        print("* " * n)
    else:
        print("* " + "  " * (n - 2) + "* ")

# 8. Butterfly Star Pattern
n = 4
for i in range(1, n + 1):
    print("* " * i + "  " * (2 * (n - i)) + "* " * i)
for i in range(n, 0, -1):
    print("* " * i + "  " * (2 * (n - i)) + "* " * i)
```

```python
# NUMBER PATTERNS.

# 1. Right-Angled Triangle Number Pattern
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

# 2. Inverted Right-Angled Number Triangle.
n = 5
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

# 3. Pyramid Number Pattern.
n = 5
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

# 4. Floyd’s Triangle.
n = 5
num = 1
for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()

# 5. Palindromic Pyramid Number Pattern.
n = 5
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, i + 1):
        print(j, end="")
    for j in range(i - 1, 0, -1):
        print(j, end="")
    print()

# 6. Pascal’s Triangle.
from math import factorial
n = 5
for i in range(n):
    print(" " * (n - i), end="")
    for j in range(i + 1):
        print(factorial(i) // (factorial(j) * factorial(i - j)), end=" ")
    print()
```

# Lists :

In Python, a list is an ordered collection of items that can hold elements of different data types. Lists are mutable, meaning their contents can be changed after creation.

**Uses :**


1.   Storing Multiple Values: Lists can store multiple related items, such as numbers, strings, or objects, in a single variable.
Example: numbers = [1, 2, 3, 4, 5]
2.   Iteration: Lists can be looped through using for loops to process each element individually.
Example: for item in my_list: print(item)
3.   Dynamic Data Handling: Lists allow adding, removing, or modifying elements, making them flexible for handling changing data.
Example: my_list.append(10) or my_list.remove(2)
4.   Nested Data Structures: Lists can contain other lists or complex objects, enabling the creation of multidimensional arrays or matrices.
Example: matrix = [1, 2], [3, 4], [5, 6]
5.   Sorting and Filtering: Lists can be sorted or filtered using built-in functions like sorted() or list comprehensions.
Example: sorted_list = sorted(my_list) or [x for x in my_list if x > 2]

```python
# A list storing different types of values
my_list = [10, 3.14, "Python", True]
print(my_list)  # Output: [10, 3.14, 'Python', True]

# Looping through a list to print each element
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# Adding and removing elements from a list
colors = ["red", "green", "blue"]
colors.append("yellow")  # Adding an element
print(colors)  # Output: ['red', 'green', 'blue', 'yellow']

colors.remove("green")  # Removing an element
print(colors)  # Output: ['red', 'blue', 'yellow']

# Creating a list of lists (2D list or matrix)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[1][2])  # Output: 6 (Accessing element in the 2nd row and 3rd column)

# Sorting a list and filtering elements using list comprehensions
numbers = [4, 2, 9, 1, 5, 6]
sorted_numbers = sorted(numbers)  # Sorting the list
print(sorted_numbers)  # Output: [1, 2, 4, 5, 6, 9]

# Filtering to get only even numbers
even_numbers = [x for x in numbers if x % 2 == 0]
print(even_numbers)  # Output: [4, 2, 6]
```

# 2D Lists
A 2D list in Python is essentially a list of lists, which can be used to represent a matrix or a table-like structure. Each element of the main list is itself a list, and these inner lists can hold multiple elements, allowing for a grid-like organization of data.

```python
# Syntax.
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```

```python
# how to create a 2D list :

# Step 1: Initialize an empty 2D list
# A 2D list is essentially a list of lists, so we begin with an empty list.
# We will fill it with rows (which are themselves lists).
matrix = []

# Step 2: Define the number of rows and columns for our 2D list
rows = 3  # We want 3 rows in our 2D list
cols = 4  # Each row will have 4 columns

# Step 3: Populate the 2D list using nested loops
# We use a loop to create each row, and within that loop, another loop to add elements to each row.

for i in range(rows):  # Loop through each row (i from 0 to 2 for 3 rows)
    # Create an empty list to represent a new row
    row = []

    for j in range(cols):  # Loop through each column in the current row
        # For simplicity, we will fill the matrix with the sum of row and column indices.
        row.append(i + j)  # Append the value (i + j) to the row

    # After filling the row, append the row to the matrix
    matrix.append(row)

# Step 4: Print the 2D list to see the result
print("The 2D list is:")
for row in matrix:  # Loop through each row in the matrix
    print(row)  # Print the row
```

```python
# Step 1: Initialize an empty 2D list
matrix = []

# Step 2: Take input for a 3x3 matrix (3 rows and 3 columns)
print("Enter the elements for a 3x3 matrix:")

for i in range(3):  # Loop through each row (3 rows)
    # Create an empty list to represent a new row
    row = []

    # Step 3: Take input for each column in the row
    for j in range(3):  # Loop through each column in the current row (3 columns)
        # Prompt the user to enter a value
        element = int(input(f"Enter element for position ({i+1}, {j+1}): "))
        # Append the entered element to the current row
        row.append(element)

    # After filling the row, append the row to the matrix
    matrix.append(row)

# Step 4: Print the 2D list (matrix)
print("\nThe 3x3 matrix you entered is:")
for row in matrix:
    print(row)
```

# Dictionaries :
In Python, a dictionary is an unordered collection of key-value pairs, where each key is unique. Keys are used to access the corresponding values, and they can be of any immutable data type (e.g., strings, numbers, tuples). Dictionaries are defined using curly braces {} with a colon separating keys and values. They are mutable, allowing the addition, removal, and modification of key-value pairs during program execution.

```python
# Creating a dictionary
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}
# Accessing dictionary values
print(person["name"])  # Output: Alice
```

```python
# Getting user input for dictionary values
name = input("Enter your name: ")
age = int(input("Enter your age: "))
city = input("Enter your city: ")
# Creating the dictionary with user input
person = {
    "name": name,
    "age": age,
    "city": city
}
# Printing the dictionary
print(person)
```

```python
# List of words
words = ["apple", "banana", "apple", "orange", "banana", "banana"]

# Dictionary to store the frequency of each word
word_count = {}

# Counting the frequency of each word
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Printing the result
print(word_count)
```

In Python, this is done by using functions, classes, and external libraries. Each module is responsible for a specific functionality, and the program can use these modules as building blocks.

# **Recursion.**
**Recursion** is a programming technique where a function calls itself in order to solve smaller instances of the same problem. It typically has two parts: the base case (to stop the recursion) and the recursive case (where the function calls itself). Recursion is useful for problems that can be broken down into similar subproblems.

```python
# Syntax :
def recursive_function(parameters):
    if base_case_condition:
        return base_case_value
    else:
        return recursive_function(smaller_problem)
```

```python
 # Recursion Example :
 """Given a number N, find all possible arrangements of N queens on an N x N chessboard where no two queens threaten each other.
You need to return all valid solutions to the N-Queens problem. Each solution will be represented by an array of integers, where the value at each index represents the column position of the queen in that row.
This is a challenging problem that involves using recursion to explore all possible configurations and backtrack when a conflict occurs."""

def solveNQueens(n):
    def isSafe(board, row, col):
        # Check this column on the current row
        for i in range(row):
            if board[i] == col or \
               board[i] - i == col - row or \
               board[i] + i == col + row:
                return False
        return True

    def solve(board, row, results):
        # Base case: If all queens are placed
        if row == n:
            results.append(["." * i + "Q" + "." * (n - i - 1) for i in board])
            return

        # Try placing the queen in all columns
        for col in range(n):
            if isSafe(board, row, col):
                board[row] = col
                solve(board, row + 1, results)  # Recur to place next queen
                board[row] = -1  # Backtrack

    results = []
    board = [-1] * n  # -1 means no queen placed
    solve(board, 0, results)
    return results

# Example usage:
n = 4
solutions = solveNQueens(n)
for solution in solutions:
    for row in solution:
        print(row)
    print()
```

# **FIBONACCI SERIES :**
**Problem:** Fibonacci Sequence with Dynamic Programming and Memoization
The Fibonacci sequence is a series of numbers in which each number is the sum of the two preceding ones. The sequence typically starts with 0 and 1. The problem is to generate the Fibonacci sequence up to the n-th term, where n is a positive integer input by the user.

**Input:**<br>
A positive integer n (1 ≤ n ≤ 10^5).


**Output:**<br>
The Fibonacci sequence up to the n-th term as a list of integers.
Also, output the time taken to compute the Fibonacci sequence for a large n (e.g., n = 100000).

```python
import time
from functools import lru_cache

# Optimized Fibonacci function using memoization with LRU cache
@lru_cache(None)
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

# Function to get the entire Fibonacci sequence up to n-th term
def fibonacci_sequence(n):
    sequence = []
    for i in range(n):
        sequence.append(fibonacci(i))
    return sequence

# Main function to run the code and measure time
def main():
    n = int(input("Enter the value of n for Fibonacci sequence: "))

    # Measure start time
    start_time = time.time()

    # Get the Fibonacci sequence up to the n-th term
    sequence = fibonacci_sequence(n)

    # Measure end time
    end_time = time.time()

    # Display results
    print(f"Fibonacci sequence up to {n}th term:")
    print(sequence)

    # Display the time taken to compute
    print(f"Time taken to compute the Fibonacci sequence up to {n}th term: {end_time - start_time} seconds")

if __name__ == "__main__":
    main()
```

# **Palindrome :**

Write a Python function longest_palindrome_substring(s: str) -> str that takes a string s as input and returns the longest contiguous palindromic substring found within it.

The input string s will consist of lowercase and/or uppercase letters.

If there are multiple palindromes of the same maximum length, return the one that appears first.

The function should be case-sensitive, so 'Aba' is not considered a palindrome.

Example:

python
Copy
Edit
Input: "babad"
Output: "bab"  # or "aba", but "bab" appears first.

```python
def longest_palindrome_substring(s: str) -> str:
    def expand_from_center(left: int, right: int) -> str:
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return s[left+1:right]

    longest = ""
    for i in range(len(s)):
        # Odd length palindrome
        temp1 = expand_from_center(i, i)
        # Even length palindrome
        temp2 = expand_from_center(i, i+1)

        if len(temp1) > len(longest):
            longest = temp1
        if len(temp2) > len(longest):
            longest = temp2

    return longest

# Example usage
print(longest_palindrome_substring("babad"))  # Output: "bab" or "aba"
print(longest_palindrome_substring("cbbd"))   # Output: "bb"
```

# **Anagram :**

You've got a troop of hyperactive monkeys typing random words on typewriters, and every time they smash the keys, they produce a random word. You suspect some of them are copying from others (naughty monkeys!), because some of the words they type are just anagrams of each other.
Your job is to find out how many groups of copycat monkeys there are — where each group consists of monkeys who typed words that are anagrams of each other.
<br><br>
**🐒 Your Task:**<br>
Write a Python function that:
Accepts a list of words typed by monkeys.
Groups the words into anagram groups.
Returns the number of unique anagram groups, along with the list of those groups.
<br>
**📥 Input:**<br>
A list of strings (e.g., ["banana", "nabana", "anaban", "apple", "papel", "hello"])
<br><br>
**📤 Output:**<br>
An integer representing the number of anagram groups.
A list of lists, where each inner list contains anagram words.
<br><br>
**🚨 Rules:**<br>
Ignore case sensitivity (e.g., "Listen" and "Silent" are anagrams).
Words with no anagrams form a group of their own.
<br><br>
**✅ Example:**<br>
python
Copy
Edit
monkey_words = ["banana", "nabana", "anaban", "apple", "papel", "hello"]
<br><br>
**Output:**<br>
css

[['banana', 'nabana', 'anaban'], ['apple', 'papel'], ['hello']]

```python
from collections import defaultdict

def find_anagram_groups(monkey_words):
    anagram_dict = defaultdict(list)

    for word in monkey_words:
        # Normalize to lowercase and sort the letters
        key = ''.join(sorted(word.lower()))
        anagram_dict[key].append(word)

    # Convert the dictionary to a list of anagram groups
    anagram_groups = list(anagram_dict.values())

    # Return number of groups and the groups themselves
    return len(anagram_groups), anagram_groups

# 🧪 Example usage
monkey_words = ["banana", "nabana", "anaban", "apple", "papel", "hello"]
num_groups, groups = find_anagram_groups(monkey_words)
print(num_groups)
print(groups)
```

# **Pangram :**
A pangram is a sentence that contains every letter of the alphabet at least once. In Python, it can be checked by comparing the set of all alphabets with the set of characters in the input sentence (after converting it to lowercase and removing non-letter characters).

**Logic :**<br>
1. Define the English alphabet (a to z).
2. Convert input sentence to lowercase.
3. Remove non-alphabet characters (optional but cleaner).
4. Create a set of characters from the sentence.
5. Check if this set contains all letters in the alphabet set.


```python
import string

def is_pangram(sentence):
    alphabet_set = set(string.ascii_lowercase)  # All 26 letters
    sentence_set = set(sentence.lower())         # Lowercase version of input
    return alphabet_set.issubset(sentence_set)   # Check if all letters are present

# Example usage
sentence = "The quick brown fox jumps over a lazy dog"
if is_pangram(sentence):
    print("Yes, it's a pangram.")
else:
    print("No, it's not a pangram.")
```

# **Armstrong Number :**
An Armstrong number (also known as a narcissistic number) is a number that is equal to the sum of its own digits each raised to the power of the number of digits.
For example, 153 is an Armstrong number because:
1³ + 5³ + 3³ = 153.

In Python programming, it's commonly used as a coding exercise to practice loops, conditionals, and math operations.

# Logic :
Take an input number (say num).<br>
Count the number of digits in the number.<br>
Split the number into individual digits.<br>
Raise each digit to the power of the total number of digits and sum them.<br>
If the sum equals the original number, it's an Armstrong number.

```python
def is_armstrong(num):
    # Convert number to string to count digits
    num_str = str(num)
    power = len(num_str)

    # Calculate sum of each digit raised to the power
    total = sum(int(digit) ** power for digit in num_str)

    # Check if it's an Armstrong number
    return num == total

# Example usage
number = 153
if is_armstrong(number):
    print(f"{number} is an Armstrong number.")
else:
    print(f"{number} is not an Armstrong number.")
```

# **Factorials :**
A factorial (denoted by n!) is the product of all positive integers from 1 to n. For example, 5! = 5 × 4 × 3 × 2 × 1 = 120.
Factorials are commonly used in permutations, combinations, and other mathematical computations.
In Python, factorials can be calculated using loops, recursion, or built-in libraries like math.factorial().

# Logic :
To calculate the factorial of a number n:

Start with a result variable set to 1.

Multiply it by each number from 1 to n (loop).

Alternatively, call a function recursively where n! = n × (n - 1)! until n is 1.


```python
#Write a Python program that takes a non-negative integer as input from the user and calculates its factorial.
# Using a loop
def factorial_iterative(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

# Using recursion
def factorial_recursive(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial_recursive(n - 1)

# Main code
num = int(input("Enter a non-negative integer: "))
if num < 0:
    print("Factorial is not defined for negative numbers.")
else:
    print(f"Factorial of {num} (Iterative): {factorial_iterative(num)}")
    print(f"Factorial of {num} (Recursive): {factorial_recursive(num)}")
```

```python
# Proper File Handling Python Example :
"""Create a program that allows a user to enter names of students and save them to a file named students.txt. The program should also be able to:
Add new student names.
View all stored student names.
Exit the program.
This simulates a simple student registration system using file handling."""

def add_student():
    name = input("Enter student name to add: ")
    with open("students.txt", "a") as file:
        file.write(name + "\n")
    print(f"'{name}' added successfully!\n")

def view_students():
    try:
        with open("students.txt", "r") as file:
            students = file.readlines()
            if students:
                print("\n--- Registered Students ---")
                for idx, student in enumerate(students, 1):
                    print(f"{idx}. {student.strip()}")
            else:
                print("No students registered yet.")
    except FileNotFoundError:
        print("No data found. Please add some students first.")

def main():
    while True:
        print("\n1. Add Student")
        print("2. View All Students")
        print("3. Exit")
        choice = input("Enter your choice: ")

        if choice == '1':
            add_student()
        elif choice == '2':
            view_students()
        elif choice == '3':
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.")

# Start the program
main()
```

# **Object Oriented Programming In Python :**

Object-Oriented Programming (OOP) in Python is a programming paradigm that organizes code using objects—which are instances of classes. A class defines a blueprint for an object, containing attributes (data) and methods (functions). Python supports core OOP concepts such as encapsulation, inheritance, polymorphism, and abstraction.

OOP is used in Python to make code more modular, reusable, and easier to maintain. It helps in structuring complex programs by dividing them into manageable pieces (objects). For example, when building a game or an app, each element (like a player, enemy, or button) can be modeled as an object with its own behavior and data. This approach promotes code reusability, simplifies debugging, and supports the DRY (Don't Repeat Yourself) principle.

```python
# 01. Classes and Objects :
"""Class: A blueprint for creating objects. It defines attributes (variables) and methods (functions).
Object: An instance of a class. It has its own values for the attributes defined in the class."""

class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, I am {self.name}")

p1 = Person("Alice")
p1.greet()  # Output: Hello, I am Alice
```

```python
# 02. Encapsulation :
""" Definition: Wrapping data (variables) and methods into a single unit (class), and restricting access to some components.
Helps in protecting the internal state of the object.
Achieved using access modifiers like _protected and __private."""

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # private variable

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount
```

```python
# 03. Inheritance :
"""Definition: One class can inherit properties and methods from another.
Promotes code reusability and supports hierarchy.
The base class is the "parent", and the derived class is the "child"."""

class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

d = Dog()
d.speak()  # Output: Dog barks
```

```python
# 04. Polymorphism :
"""Definition: The same method name can have different behavior based on the object.
Can be achieved by method overriding (in inheritance) or method overloading (by default, Python supports overriding)."""

class Cat:
    def sound(self):
        print("Meow")

class Cow:
    def sound(self):
        print("Moo")

# Polymorphism in action
def make_sound(animal):
    animal.sound()

make_sound(Cat())  # Meow
make_sound(Cow())  # Moo
```

```python
# 05. Abstraction :
"""Definition: Hiding complex internal details and showing only essential features.
Achieved using abstract base classes with abc module."""

from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car started")

c = Car()
c.start()  # Output: Car started
```

```python
# Python Code where Advanced Data stuructures are used :
from collections import Counter
import heapq

def top_k_frequent_words(text, k):
    # Step 1: Tokenize the text
    words = text.lower().split()

    # Step 2: Count word frequency
    freq = Counter(words)

    # Step 3: Use a heap to get top k frequent
    heap = [(-count, word) for word, count in freq.items()]  # Max-heap by negating count
    heapq.heapify(heap)

    # Step 4: Extract top k
    top_k = [heapq.heappop(heap)[1] for _ in range(k)]

    return top_k

# Example usage
text = "apple banana apple orange banana apple mango banana"
print(top_k_frequent_words(text, 3))  # Output: ['apple', 'banana', 'orange']
```
