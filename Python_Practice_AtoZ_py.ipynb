```python
# Learning Printing Statements.

# Method 1
print("Hello World!!")
# Method 2
age= 21
roll_no = "DY22ENGUOAID023"
print("My Name is Atharva Ghayal")
print("I am ", age, " years old")
print("my college roll.no is :",roll_no)
# Method 3
print("Atharva","Chandrashekhar","Ghayal",sep="|")
# Method 4
print("Hello",end=" ")
print("World!")
# Method 5
name = "Atharva"
age = 21
print("My name is {} and i am {} years old!!".format(name,age))
# Method 6
print(f"My name is {name} and i am {age} years old.")
# Method 7(escape sequences.)
print("Hello\nWorld!!")
print("Hello\tWorld!!")
print("My name is,\"Atharva Ghayal!\"")
# Method 8(raw text if dont want escape sequences to be used)
value = "Hello\nworld"
print(repr(value))
# Method 9
import sys
sys.stdout.write("Hello World!!\n")
# Method 10
with open("output.txt","w") as file:
  print("this to be written into the file accordingly!!", file=file)
# Method 11
print("Unicode Heart: \u2764")
print("Smiley Emoji: \U0001F600")
# Method 12
numbers = [1, 2, 3, 4, 5]
print(*numbers)  # Unpacks the list
# Method 13
print("H", "E", "L", "L", "O", sep="")
#Method 14
num = 255
print("Hexadecimal:", hex(num))
print("Binary:", bin(num))
print("Octal:", oct(num))
# Method 15
import time
for i in range(5):
    print(f"Processing {i}...", end="\r", flush=True)
    time.sleep(1)
# Method 16
words = ["Python", "is", "awesome"]
print(" ".join(words))
# Method 17
import json
data = {"name": "Atharva", "age": 21, "language": "Python"}
print(json.dumps(data, indent=4))
# Method 18
from pprint import pprint
data = [{"name": "Atharva", "skills": ["Python", "Flask", "Java"]}, {"name": "Raj", "skills": ["C++", "JS"]}]
pprint(data)
# Method 19
from tabulate import tabulate
data = [["Atharva", 21, "Python"], ["Raj", 22, "Java"]]
print(tabulate(data, headers=["Name", "Age", "Language"], tablefmt="grid"))
# Method 20
from tqdm import tqdm
import time

for _ in tqdm(range(10), desc="Loading..."):
    time.sleep(0.5)
```
