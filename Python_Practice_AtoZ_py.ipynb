## **Control statements** in Python are used to manage the flow of execution in a program. They help in decision-making, looping, and altering the normal sequence of execution.

**Conditional Statements** (if, elif, else) – These are used to execute a block of code based on specific conditions.

**Looping Statements** (for, while) – These allow repeated execution of a block of code until a condition is met.

**Jump Statements** (break, continue, pass) – break exits a loop, continue skips the current iteration, and pass acts as a placeholder.

These statements help in writing efficient and structured programs.

```python
# Control Statements :

# Conditional Statements :

age = int(input("Enter your age: "))

if age < 18:
    print("You are a minor.")
elif 18 <= age < 60:
    print("You are an adult.")
else:
    print("You are a senior citizen.")

# Looping Statement(using for loop) :

for i in range(1, 6):  # Loops from 1 to 5
    print(f"Number: {i}")

#Looping Statement(using while loop) :

count = 1
while count <= 5:
    print(f"Count: {count}")
    count += 1

# Jump Statements (break,pass,continue) :

for num in range(1, 6):
    if num == 3:
        continue  # Skips 3
    elif num == 5:
        break  # Stops loop at 5
    else:
        pass  # Placeholder (does nothing)
    print(num)

# All (Control statement,Looping Statement, Jump Statement Combined Example) :

for i in range(1, 11):
    if i % 2 == 0:
        continue  # Skips even numbers
    elif i == 9:
        break  # Stops loop when `i` is 9
    else:
        print(f"Odd Number: {i}")  # Prints only odd numbers except 9
```

```python
# Fully Combined Example :

import random

# Generate a random number between 1 and 20
secret_number = random.randint(1, 20)
attempts = 0

print("Welcome to the Number Guessing Game! (Guess between 1-20)")

# Using a for loop to allow a maximum of 5 attempts
for attempt in range(1, 6):
    guess = int(input(f"Attempt {attempt}: Enter your guess: "))
    attempts += 1

    if guess < secret_number:
        print("Too low! Try again.")
        continue  # Skips the remaining part of the loop and starts next iteration
    elif guess > secret_number:
        print("Too high! Try again.")
        pass  # Does nothing, just a placeholder
    else:
        print(f"🎉 Congratulations! You guessed it in {attempts} attempts.")
        break  # Exits loop if guessed correctly

# If the user couldn't guess in 5 attempts, allow them an extra chance using `while` loop
else:
    print("\nOops! You've used all 5 attempts.")

    # Asking for an extra attempt
    while True:
        choice = input("Do you want one last chance? (yes/no): ").lower()

        if choice == "yes":
            extra_guess = int(input("Enter your final guess: "))

            if extra_guess == secret_number:
                print(f"🎉 Great! You guessed it right in the extra attempt!")
            else:
                print(f"❌ Sorry! The correct number was {secret_number}.")
            break  # Exits while loop
        elif choice == "no":
            print(f"The correct number was {secret_number}. Better luck next time!")
            break
        else:
            print("Invalid input, please type 'yes' or 'no'.")
```

# Nested Loop :
A nested loop is a loop inside another loop. The inner loop executes completely for each iteration of the outer loop.

```python
for i in range(4):  # 4 rows
    for j in range(4):  # 4 columns
        print("#", end=" ")
    print()  # Move to the next line
```

```python
for i in range(3):  # Outer loop runs 3 times
    for j in range(2):  # Inner loop runs 2 times for each outer loop iteration
        print(f"i = {i}, j = {j}")
```

# Nested If, ElseIf, Else Statements :
In Python, nested if, elif, else statements allow multiple levels of decision-making. A nested if means placing one if inside another, helping to check multiple conditions sequentially. If the outer if condition is true, the inner if-elif-else block executes. This is useful when decision-making depends on multiple conditions being met in a specific order.

```python
# Checking Eligibility for Loan Based on Salary & Credit Score :
salary = int(input("Enter your monthly salary: "))
credit_score = int(input("Enter your credit score: "))

if salary >= 50000:
    if credit_score >= 750:
        print("Eligible for High Loan Amount")
    else:
        print("Improve your credit score for a high loan amount")
elif salary >= 30000:
    if credit_score >= 650:
        print("Eligible for Medium Loan Amount")
    else:
        print("Improve your credit score for a medium loan amount")
elif salary >= 20000:
    if credit_score >= 550:
        print("Eligible for Low Loan Amount")
    else:
        print("Improve your credit score for a low loan amount")
else:
    print("Not Eligible for Loan")
```

# Star and Number Patterns in Python :

Pattern printing is one of the fundamental concepts in Python, often used to develop logic-building skills for beginners. It involves using nested loops (one loop controlling rows and another controlling columns) to display various patterns such as stars (*), numbers, alphabets, or special characters. These patterns help in understanding loops, conditionals, and iterations, which are essential for programming logic.

Star and number patterns are widely used in competitive programming, logic development, and technical interviews to test problem-solving ability. They also find applications in graphics programming, ASCII art, and mathematical representations in Python. Below are Python programs for all major types of star patterns and number patterns.

```python
# Python Programs for Star Patterns...

# 1. for Simple Right-Angled Triangle Star Pattern..
n = 5
for i in range(1, n + 1):
    print("* " * i)

# 2.Inverted Right-Angled Triangle Star Pattern.
n = 5
for i in range(n, 0, -1):
    print("* " * i)

# 3. Pyramid Star Pattern.
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)

# 4. Inverted Pyramid Star Pattern.
n = 5
for i in range(n, 0, -1):
    print(" " * (n - i) + "* " * i)

# 5. Diamond Star Pattern
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " * i)

# 6. Hollow Diamond Star Pattern.
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "* " if i == 1 else " " * (n - i) + "* " + "  " * (i - 2) + "* ")
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " if i == 1 else " " * (n - i) + "* " + "  " * (i - 2) + "* ")

# 7. Hollow Square Star Pattern.
n = 5
for i in range(n):
    if i == 0 or i == n - 1:
        print("* " * n)
    else:
        print("* " + "  " * (n - 2) + "* ")

# 8. Butterfly Star Pattern
n = 4
for i in range(1, n + 1):
    print("* " * i + "  " * (2 * (n - i)) + "* " * i)
for i in range(n, 0, -1):
    print("* " * i + "  " * (2 * (n - i)) + "* " * i)
```

```python
# NUMBER PATTERNS.

# 1. Right-Angled Triangle Number Pattern
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

# 2. Inverted Right-Angled Number Triangle.
n = 5
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

# 3. Pyramid Number Pattern.
n = 5
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

# 4. Floyd’s Triangle.
n = 5
num = 1
for i in range(1, n + 1):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()

# 5. Palindromic Pyramid Number Pattern.
n = 5
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, i + 1):
        print(j, end="")
    for j in range(i - 1, 0, -1):
        print(j, end="")
    print()

# 6. Pascal’s Triangle.
from math import factorial
n = 5
for i in range(n):
    print(" " * (n - i), end="")
    for j in range(i + 1):
        print(factorial(i) // (factorial(j) * factorial(i - j)), end=" ")
    print()
```

# Lists :

In Python, a list is an ordered collection of items that can hold elements of different data types. Lists are mutable, meaning their contents can be changed after creation.

**Uses :**


1.   Storing Multiple Values: Lists can store multiple related items, such as numbers, strings, or objects, in a single variable.
Example: numbers = [1, 2, 3, 4, 5]
2.   Iteration: Lists can be looped through using for loops to process each element individually.
Example: for item in my_list: print(item)
3.   Dynamic Data Handling: Lists allow adding, removing, or modifying elements, making them flexible for handling changing data.
Example: my_list.append(10) or my_list.remove(2)
4.   Nested Data Structures: Lists can contain other lists or complex objects, enabling the creation of multidimensional arrays or matrices.
Example: matrix = [1, 2], [3, 4], [5, 6]
5.   Sorting and Filtering: Lists can be sorted or filtered using built-in functions like sorted() or list comprehensions.
Example: sorted_list = sorted(my_list) or [x for x in my_list if x > 2]

```python
# A list storing different types of values
my_list = [10, 3.14, "Python", True]
print(my_list)  # Output: [10, 3.14, 'Python', True]

# Looping through a list to print each element
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

# Adding and removing elements from a list
colors = ["red", "green", "blue"]
colors.append("yellow")  # Adding an element
print(colors)  # Output: ['red', 'green', 'blue', 'yellow']

colors.remove("green")  # Removing an element
print(colors)  # Output: ['red', 'blue', 'yellow']

# Creating a list of lists (2D list or matrix)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[1][2])  # Output: 6 (Accessing element in the 2nd row and 3rd column)

# Sorting a list and filtering elements using list comprehensions
numbers = [4, 2, 9, 1, 5, 6]
sorted_numbers = sorted(numbers)  # Sorting the list
print(sorted_numbers)  # Output: [1, 2, 4, 5, 6, 9]

# Filtering to get only even numbers
even_numbers = [x for x in numbers if x % 2 == 0]
print(even_numbers)  # Output: [4, 2, 6]
```

# 2D Lists
A 2D list in Python is essentially a list of lists, which can be used to represent a matrix or a table-like structure. Each element of the main list is itself a list, and these inner lists can hold multiple elements, allowing for a grid-like organization of data.

```python
# Syntax.
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```

```python
# how to create a 2D list :

# Step 1: Initialize an empty 2D list
# A 2D list is essentially a list of lists, so we begin with an empty list.
# We will fill it with rows (which are themselves lists).
matrix = []

# Step 2: Define the number of rows and columns for our 2D list
rows = 3  # We want 3 rows in our 2D list
cols = 4  # Each row will have 4 columns

# Step 3: Populate the 2D list using nested loops
# We use a loop to create each row, and within that loop, another loop to add elements to each row.

for i in range(rows):  # Loop through each row (i from 0 to 2 for 3 rows)
    # Create an empty list to represent a new row
    row = []

    for j in range(cols):  # Loop through each column in the current row
        # For simplicity, we will fill the matrix with the sum of row and column indices.
        row.append(i + j)  # Append the value (i + j) to the row

    # After filling the row, append the row to the matrix
    matrix.append(row)

# Step 4: Print the 2D list to see the result
print("The 2D list is:")
for row in matrix:  # Loop through each row in the matrix
    print(row)  # Print the row
```

```python
# Step 1: Initialize an empty 2D list
matrix = []

# Step 2: Take input for a 3x3 matrix (3 rows and 3 columns)
print("Enter the elements for a 3x3 matrix:")

for i in range(3):  # Loop through each row (3 rows)
    # Create an empty list to represent a new row
    row = []

    # Step 3: Take input for each column in the row
    for j in range(3):  # Loop through each column in the current row (3 columns)
        # Prompt the user to enter a value
        element = int(input(f"Enter element for position ({i+1}, {j+1}): "))
        # Append the entered element to the current row
        row.append(element)

    # After filling the row, append the row to the matrix
    matrix.append(row)

# Step 4: Print the 2D list (matrix)
print("\nThe 3x3 matrix you entered is:")
for row in matrix:
    print(row)
```
